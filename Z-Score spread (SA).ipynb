```python
# prompt: create frontend for upper code give button and esy to use perfect interface to change ticker name and all

import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# Function to perform the analysis and plot
def analyze_spread(ticker1, ticker2, start_date, end_date):
    try:
        # Download historical data
        data = yf.download([ticker1, ticker2], start=start_date, end=end_date)

        # Calculate spread, Z-score, and signals (same as before)
        data['Spread'] = data['Close'][ticker1] - data['Close'][ticker2]
        data['Mean'] = data['Spread'].rolling(window=30).mean()
        data['Std'] = data['Spread'].rolling(window=30).std()
        data['Z-Score'] = (data['Spread'] - data['Mean']) / data['Std']

        data['Signal'] = 0
        data['Signal'][data['Z-Score'] < -2] = 1
        data['Signal'][data['Z-Score'] > 2] = -1


        # Plotting
        plt.figure(figsize=(14, 7))
        plt.plot(data.index, data['Z-Score'], label='Z-Score', color='blue')
        plt.axhline(2, color='red', linestyle='--', label='Upper Threshold (2)')
        plt.axhline(-2, color='green', linestyle='--', label='Lower Threshold (-2)')
        plt.title(f'Z-Score of {ticker1} and {ticker2} Spread')
        plt.xlabel('Date')
        plt.ylabel('Z-Score')
        plt.legend()
        plt.grid()
        plt.show()

        plt.figure(figsize=(14, 7))
        plt.plot(data.index, data['Spread'], label='Price Spread', color='blue')
        plt.scatter(data.index[data['Signal'] == 1], data['Spread'][data['Signal'] == 1], marker='^', color='green', label='Buy Signal', s=100)
        plt.scatter(data.index[data['Signal'] == -1], data['Spread'][data['Signal'] == -1], marker='v', color='red', label='Sell Signal', s=100)
        plt.title(f'Price Spread of {ticker1} and {ticker2} with Trading Signals')
        plt.xlabel('Date')
        plt.ylabel('Price Spread')
        plt.legend()
        plt.grid()
        plt.show()
    except Exception as e:
        print(f"An error occurred: {e}")

# Create widgets
ticker1_widget = widgets.Text(value='AAPL', description='Ticker 1:')
ticker2_widget = widgets.Text(value='MSFT', description='Ticker 2:')
start_date_widget = widgets.DatePicker(description='Start Date:', value=pd.to_datetime('2021-01-01'))
end_date_widget = widgets.DatePicker(description='End Date:', value=pd.to_datetime('2021-12-31'))
analyze_button = widgets.Button(description='Analyze')


# Output widget to display plots
output = widgets.Output()

# Function to handle button click
def on_button_clicked(b):
    with output:
        clear_output()
        analyze_spread(ticker1_widget.value, ticker2_widget.value, start_date_widget.value, end_date_widget.value)
analyze_button.on_click(on_button_clicked)

# Display widgets
display(ticker1_widget, ticker2_widget, start_date_widget, end_date_widget, analyze_button, output)
```
